```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/content/Retail_Dataset_Cleaned.csv")

print("Dataset Loaded Successfully")
```

```python
df.head()
```

```python
df.info()
```

```python
segment_revenue = pd.pivot_table(
    df,
    values="Revenue",
    index="Customer_Segment",
    aggfunc="sum"
)

segment_revenue
```

```python
category_summary = pd.pivot_table(
    df,
    values="Revenue",
    index="Category",
    aggfunc=["sum", "mean", "max", "min"]
)
category_summary
```

```python
channel_summary = pd.pivot_table(
    df,
    values="Revenue",
    index="Sales_Channel",
    aggfunc=["sum", "mean"]
)
channel_summary
```

```python
city_performance = pd.pivot_table(
    df,
    values=["Revenue", "Profit_Margin"],
    index="City",
    aggfunc={
        "Revenue": "sum",
        "Profit_Margin": "mean"
    }
)
city_performance
```

```python
store_summary = pd.pivot_table(
    df,
    values=["Revenue", "Profit_Margin"],
    index="Store_Type",
    aggfunc="sum"
)
store_summary
```

```python
store_name = pd.pivot_table(
    df,
    values=["Profit_Margin"],
    index=["Store_Name"],
    aggfunc="sum"
)
store_name
```

```python
product_name = pd.pivot_table(
    df,
    values=["Quantity_Available"],
    index=["Product_Name"],
    aggfunc="sum"
)
product_name
```
